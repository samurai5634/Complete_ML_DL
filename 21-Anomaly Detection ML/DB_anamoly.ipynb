## DBSCAN Clustering

### DBSCAN is specifically used for non-linear data and to find outliers with this is a piece of cake. Because after segregation it is very visible, as we know there are three kinds of points -:
1. Core
2. Border
3. Outlier(with values -1) 

```python
from sklearn.cluster import DBSCAN
```

```python

import matplotlib.pyplot as plt
%matplotlib inline
from sklearn.datasets import make_circles
```

```python

X, y = make_circles(n_samples=750,factor=0.5,noise=0.05)
```

```python
X
```

```python
plt.scatter(X[:,0],X[:,1],cmap='magma')
```

```python
y
```

```python
from sklearn.cluster import DBSCAN
```

```python
dbcan=DBSCAN(eps=0.1)
```

```python
dbcan.fit_predict(X)
```

```python
val = dbcan.labels_
```

```python
import numpy as np
uniq,count = np.unique(val,return_counts=True) 
```

```python
num_groups = len(uniq)
print("Unique Groups/Values:", uniq)  # Output: [0 1]
print("Counts per Group:    ", count)         # Output: [4 3]
print("Total No. of Groups: ", num_groups)     # Output: 2
```

```python
plt.scatter(X[:,0],X[:,1],c=dbcan.labels_,edgecolors='r')
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
```
